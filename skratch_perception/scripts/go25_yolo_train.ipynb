```python
import yaml

attr = {
    # 'path': cwd,
    'train': './../lego/train',
    'val': './../lego/val',
    'nc': 8,
    'names': {
                0: "2x2_blue_block",
                1: "2x2_green_block",
                2: "2x2_red_block",
                3: "2x2_yellow_block",
                4: "4x2_blue_block",
                5: "4x2_green_block",
                6: "4x2_red_block",
                7: "4x2_yellow_block"
}}
```

```python
from ultralytics import YOLO

with open('./../config/object_names.yaml', 'w') as f:
    yaml.dump(attr, f)
```

```python
from ultralytics import YOLO

model = YOLO('yolo11n-obb.pt')
data_path = './../config/object_names.yaml'
trained_path = './../lego/trained_models'
name_path = 'legoresults_'

results = model.train(data=data_path, epochs=2500, imgsz=(480, 640), patience=250, project=trained_path, name=name_path)
```

```python
trained_model_path = os.path.join(trained_path, name_path + '2', 'weights', 'best.pt')
test_path = './../lego/test/images'
prediction_path = './../lego/trained_predictions'

! yolo obb predict model=$trained_model_path source=$test_path name=$prediction_path exist_ok=True
```
