```python
import os
import json
from ultralytics import YOLO

# Load trained YOLO model
model_path = "/home/bitbots/Downloads/rcaw25/dataset/rcaw25_day_3/model/best_empty.pt"
model = YOLO(model_path)

# Define input and output paths
# Define paths
image_folder = "/home/bitbots/Downloads/rcaw25/dataset/rcaw25_day_3/new_empty_space_images"  # Folder with 200 images
output_json_folder = "/home/bitbots/Downloads/rcaw25/dataset/rcaw25_day_3/new_json_labels"  # Folder to save JSONs


# Ensure output folder exists
os.makedirs(output_json_folder, exist_ok=True)

# Get all image files
image_files = sorted([f for f in os.listdir(image_folder) if f.endswith((".jpg", ".png", ".jpeg"))])

# Run inference on all images
results = model([os.path.join(image_folder, img) for img in image_files])

# Convert predictions to the required JSON format
for img_name, result in zip(image_files, results):
    json_filename = os.path.join(output_json_folder, f"{os.path.splitext(img_name)[0]}.json")

    # Define the JSON structure
    json_data = {
        "version": "2.4.1",
        "flags": {},
        "shapes": [],
        "imagePath": img_name,
        "imageData": None,
        "imageHeight": 480,  # Update this dynamically if needed
        "imageWidth": 640,   # Update this dynamically if needed
        "description": ""
    }

    # Process each detected bounding boxd
    for box, cls in zip(result.boxes.xyxy, result.boxes.cls):
        x1, y1, x2, y2 = box.tolist()  # Convert to list format
        category = "empty_space"  # Use your class label (assuming single-class)

        # Convert bbox to the required "points" format (clockwise order)
        points = [
            [x1, y1],  # Top-left
            [x2, y1],  # Top-right
            [x2, y2],  # Bottom-right
            [x1, y2]   # Bottom-left
        ]

        # Append the annotation
        json_data["shapes"].append({
            "kie_linking": [],
            "direction": 0,
            "label": category,
            "score": None,
            "points": points,
            "group_id": None,
            "description": "",
            "difficult": False,
            "shape_type": "rotation",  # If rotation is incorrect, change to "rectangle"
            "flags": {},
            "attributes": {}
        })

    # Save JSON file
    with open(json_filename, "w") as json_file:
        json.dump(json_data, json_file, indent=4)

    print(f"✅ Saved: {json_filename}")

print("\n🚀 JSON files successfully generated in the required format!")
```
